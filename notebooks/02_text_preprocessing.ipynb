# Notebook 02: Text Preprocessing

In this notebook, we clean the raw movie reviews and save a cleaned dataset for model training.

```python
import sys
import os

sys.path.append(os.path.abspath(".."))

import pandas as pd

from utils import load_data, clean_text
```

```python
df = load_data("../data/imdb_reviews.csv")
df.head()
```

```python
df["review"].iloc[0]
```

```python
sample_review = df["review"].iloc[0]

print("Original Review:")
print(sample_review[:1000])

print("\nCleaned Review:")
print(clean_text(sample_review)[:1000])
```

```python
df["clean_review"] = df["review"].apply(clean_text)

df.head()
```

```python
df[["review", "clean_review", "sentiment"]].head()
```

```python
empty_reviews = df[df["clean_review"].str.strip() == ""]
empty_reviews.shape
```

```python
df = df[df["clean_review"].str.strip() != ""]
df.shape
```

```python
df.to_csv("../data/imdb_cleaned.csv", index=False)

print("Cleaned dataset saved successfully!")
```

## Text Preprocessing Summary

The following cleaning steps were applied:

1. Converted text to lowercase
2. Removed HTML tags
3. Removed URLs
4. Removed numbers, punctuation, and special characters
5. Removed stopwords
6. Kept important negation words such as `not`, `no`, and `never`
7. Applied stemming
8. Saved the cleaned dataset as `imdb_cleaned.csv`

The cleaned dataset will be used for model training.
